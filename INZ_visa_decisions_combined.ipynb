```python
################################################################################################################################################################
## Column Appender # `1556`
# Node state: `EXECUTED`
# Input: [Port 1] 1547:1 from Joiner #1547; [Port 2] 1551:1 from Joiner #1551; [Port 3] 1592:1 from Joiner #1592
# Output: [Port 1] 1556:1 to 1561 Missing Value #1561
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.columnappend2.ColumnAppender2NodeFactory
left_df = context['1547:1']  # left table
right_dfs = []
right_dfs.append(context['1551:1'])
right_dfs.append(context['1592:1'])
out_df = left_df.copy()
_mode = 'IDENTICAL'

def _unique_with_suffix(base: str, existing: set, suffix: str) -> str:
    if base not in existing:
        return base
    new = f"{base}{suffix}"
    # If even the suffixed name exists, keep appending the same suffix
    while new in existing:
        new = f"{new}{suffix}"
    return new

# Append each right table in order (1-based index for suffix)
for k, rdf in enumerate(right_dfs, start=1):
    suffix = f" (#{k})"
    # Build a collision-aware rename map guaranteeing uniqueness
    existing = set(out_df.columns)
    rename_map = {}
    for c in rdf.columns:
        if c in existing or c in rename_map.values():
            newc = _unique_with_suffix(c, existing | set(rename_map.values()), suffix)
            rename_map[c] = newc
        else:
            # no collision; keep original
            rename_map[c] = c
    right_safe = rdf.rename(columns=rename_map) if rename_map else rdf

    if _mode == 'IDENTICAL':
        # Align by index; if different, left-align by reindexing right to out_df.index
        if out_df.index.equals(right_safe.index):
            out_df = out_df.join(right_safe, how='left')
        else:
            out_df = out_df.join(right_safe.reindex(out_df.index), how='left')
    else:
        # Positional fallback for unknown modes
        out_df = pd.concat([out_df.reset_index(drop=True),
                           right_safe.reset_index(drop=True)], axis=1)
context['1556:1'] = out_df
```

```python
################################################################################################################################################################
## Missing Value # `1561`
# Node state: `EXECUTED`
# Input: [Port 1] 1556:1 from Column Appender #1556
# Output: [Port 1] 1561:1 to 1557 Column Filter #1557
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.pmml.missingval.compute.MissingValueHandlerNodeFactory
df = context['1556:1']  # input table
def _mv_dtype_key(series):
    if pd.api.types.is_integer_dtype(series):
        return 'int'
    if pd.api.types.is_float_dtype(series):
        return 'float'
    if pd.api.types.is_bool_dtype(series):
        return 'boolean'
    return 'string'

def _mv_policy_for(col_name, dtype_key, column_policies, dtype_policies):
    for pol in column_policies:
        if pol.get('column') == col_name:
            return pol
    for pol in dtype_policies:
        if pol.get('dtype') == dtype_key:
            return pol
    return None

def _mv_resolve_value(series, policy, dtype_key):
    if policy is None:
        return None
    strat = (policy.get('strategy') or '').lower()
    if strat == 'fixed':
        val = policy.get('value')
    elif strat == 'mean':
        val = series.mean(skipna=True)
    elif strat == 'median':
        val = series.median(skipna=True)
    elif strat == 'mode':
        mode = series.mode(dropna=True)
        val = mode.iloc[0] if not mode.empty else None
    else:
        val = policy.get('value')
    if val is None or (isinstance(val, float) and pd.isna(val)):
        return None
    if dtype_key == 'int':
        try:
            val = int(round(float(val)))
        except Exception:
            return None
        return str(val), 'integer'
    if dtype_key == 'float':
        try:
            val = float(val)
        except Exception:
            return None
        return str(val), 'double'
    if dtype_key == 'boolean':
        text = str(val).strip().lower()
        val_int = '1' if text in {'true','1','t','y','yes'} else '0'
        return val_int, 'integer'
    return str(val), 'string'

def _mv_collect_bundle(source_df, column_policies, dtype_policies):
    df = source_df
    data_entries = []
    transforms = []
    for col in df.columns:
        series = df[col]
        dtype_key = _mv_dtype_key(series)
        entry = {'name': str(col), 'dtype_key': dtype_key}
        if dtype_key == 'string':
            entry['optype'] = 'categorical'
            entry['dataType'] = 'string'
            values = series.dropna().astype(str).unique().tolist()
            if values:
                entry['values'] = [str(v) for v in values[:128]]
        else:
            entry['optype'] = 'continuous'
            entry['dataType'] = 'integer' if dtype_key in {'int','boolean'} else 'double'
            non_missing = series.dropna()
            if not non_missing.empty:
                try:
                    entry['interval'] = [float(non_missing.min()), float(non_missing.max())]
                except Exception:
                    pass
        data_entries.append(entry)
        policy = _mv_policy_for(col, dtype_key, column_policies, dtype_policies)
        resolved = _mv_resolve_value(series, policy, dtype_key)
        if resolved is None:
            continue
        const_text, const_dtype = resolved
        transforms.append({
            'column': str(col),
            'derived_name': f"{col}*",
            'const': const_text,
            'const_dtype': const_dtype,
            'optype': entry.get('optype', 'continuous'),
        })
    return {
        'model_type': 'missing_value',
        'version': '4.2',
        'application': {'name': 'knime2py', 'version': '1.0'},
        'data_dictionary': data_entries,
        'transformations': transforms,
    }
out_df = df.copy()
override_cols = {'GDP_per_capita_current_USD', 'WB_FY2026_Income_Group', 'WB_FY2026_Lending_Category', 'WHO_DTP3_Immunization_Coverage_pct', 'WHO_Life_Expectancy_at_Birth_years', 'WHO_Physician_Density_per_10000_population'}
if 'WHO_Life_Expectancy_at_Birth_years' in out_df.columns:
    out_df = out_df.dropna(subset=['WHO_Life_Expectancy_at_Birth_years'])
if 'WHO_DTP3_Immunization_Coverage_pct' in out_df.columns:
    out_df = out_df.dropna(subset=['WHO_DTP3_Immunization_Coverage_pct'])
if 'GDP_per_capita_current_USD' in out_df.columns:
    out_df = out_df.dropna(subset=['GDP_per_capita_current_USD'])
if 'WB_FY2026_Income_Group' in out_df.columns:
    out_df['WB_FY2026_Income_Group'] = out_df['WB_FY2026_Income_Group'].fillna('Missing')
if 'WB_FY2026_Lending_Category' in out_df.columns:
    out_df['WB_FY2026_Lending_Category'] = out_df['WB_FY2026_Lending_Category'].fillna('Missing')
if 'WHO_Physician_Density_per_10000_population' in out_df.columns:
    out_df = out_df.dropna(subset=['WHO_Physician_Density_per_10000_population'])
int_cols = [c for c in out_df.select_dtypes(include=['Int64','Int32','Int16','int64','int32','int16']).columns if c not in override_cols]
if len(int_cols) > 0:
    pass  # No fixed value configured for ints; skipping
float_cols = [c for c in out_df.select_dtypes(include=['float64','float32']).columns if c not in override_cols]
if len(float_cols) > 0:
    pass  # No fixed value configured for floats; skipping
str_cols = [c for c in out_df.select_dtypes(include=['string','object']).columns if c not in override_cols]
if len(str_cols) > 0:
    pass  # No fixed value configured for strings; skipping
dtype_policies = [{'dtype': 'int', 'strategy': 'fixed', 'value': None}, {'dtype': 'string', 'strategy': 'fixed', 'value': None}, {'dtype': 'float', 'strategy': 'fixed', 'value': None}]
column_policies = [{'column': 'WHO_Life_Expectancy_at_Birth_years', 'dtype': None, 'strategy': 'drop', 'value': None}, {'column': 'WHO_DTP3_Immunization_Coverage_pct', 'dtype': None, 'strategy': 'drop', 'value': None}, {'column': 'GDP_per_capita_current_USD', 'dtype': None, 'strategy': 'drop', 'value': None}, {'column': 'WB_FY2026_Income_Group', 'dtype': 'string', 'strategy': 'fixed', 'value': 'Missing'}, {'column': 'WB_FY2026_Lending_Category', 'dtype': 'string', 'strategy': 'fixed', 'value': 'Missing'}, {'column': 'WHO_Physician_Density_per_10000_population', 'dtype': None, 'strategy': 'drop', 'value': None}]
model_bundle = _mv_collect_bundle(df, column_policies, dtype_policies)
model_bundle['strategies'] = dtype_policies
model_bundle['column_strategies'] = column_policies
context['1561:1'] = out_df
```

## Linear Correlation \# `1552`
Node state: `EXECUTED`  
Input: [Port 1] 1557:1 from Column Filter #1557  
Output: [Port 1] 1552:1 to 1593 CSV Writer #1593  


```python
################################################################################################################################################################
## Linear Correlation # `1552`
# Node state: `EXECUTED`
# Input: [Port 1] 1557:1 from Column Filter #1557
# Output: [Port 1] 1552:1 to 1593 CSV Writer #1593
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.correlation.compute2.CorrelationCompute2NodeFactory
df = context['1557:1']  # input table
out_df = df.copy()  # passthrough copy

# 1) Type buckets and eligible nominal limit
_pvc = int(50)
_num_like_all = df.select_dtypes(include=['number','Int64','Float64','bool','boolean']).columns.tolist()
_nom_cand_all = df.select_dtypes(include=['object','string','category']).columns.tolist()
_nom_eligible_all = []
for _c in _nom_cand_all:
    _k = int(df[_c].astype('string').nunique(dropna=True))
    if _k <= _pvc:
        _nom_eligible_all.append(_c)

# 2) Establish the SOURCE ORDER
_included_cfg = ['Approval rate', 'WHO_Under5_Mortality_per_1000_live_births', 'WHO_Maternal_Mortality_per_100000_live_births', 'WHO_Physician_Density_per_10000_population', 'WHO_DTP3_Immunization_Coverage_pct', 'GDP_per_capita_current_USD', 'Life_expectancy_at_birth_years', 'Internet_users_pct_population', 'Urban_population_pct', 'GNI_per_capita_Atlas_current_USD']
_excluded_cfg = ['Country', 'Applications_log10', 'WHO_Region', 'WB_FY2026_Region', 'WB_FY2026_Income_Group', 'WB_FY2026_Lending_Category', 'UN M49']
_enforce = 'EnforceExclusion'
if _enforce == 'EnforceInclusion' and len(_included_cfg) > 0:
    # Exact order from settings.xml (filtered to existing columns)
    _order_source = [c for c in _included_cfg if c in out_df.columns]
else:
    # Input table order
    _order_source = list(df.columns)

# 3) Apply include/exclude and compatibility (no reordering)
if _enforce == 'EnforceInclusion' and len(_included_cfg) > 0:
    _union = [c for c in _order_source if c in _included_cfg]
else:  # EnforceExclusion
    _union = [c for c in _order_source if c not in set(_excluded_cfg)]

# Buckets in SOURCE order
_num_like = [c for c in _union if c in _num_like_all]
_nom_eligible = [c for c in _union if c in _nom_eligible_all]

# 4) Build Port-1 measure rows in SOURCE order
_pairs_mode = 'COMPATIBLE_PAIRS'  # 'COMPATIBLE_PAIRS' | 'ALL_PAIRS'
pairs = []
for i in range(len(_union)):
    for j in range(i+1, len(_union)):
        a, b = _union[i], _union[j]
        if _pairs_mode == 'COMPATIBLE_PAIRS':
            # Keep order; only add if both numeric or both eligible nominal
            if (a in _num_like and b in _num_like) or (a in _nom_eligible and b in _nom_eligible):
                pairs.append((a, b))
        else:  # ALL_PAIRS
            pairs.append((a, b))

# Early-empty check
if len(pairs) == 0:
    measure_out = pd.DataFrame(columns=[
        'First column name', 'Second column name', 'Correlation value', 'p value', 'Degrees of freedom'
    ])
else:
    try:
        _have_scipy = True
    except Exception:
        _have_scipy = False
    _alt = 'TWO_SIDED'

    def _encode_series(s):
        # Nominal → ordered integers with NaN preserved; numeric/bool → float64
        if s.dtype.name in ('object','string') or str(s.dtype).startswith('category'):
            _u = pd.Series(sorted(s.dropna().astype('string').unique()), dtype='string')
            _map = {v:i for i,v in enumerate(_u.tolist())}
            _enc = s.astype('string').map(_map)
            return _enc.astype('float64')
        return pd.to_numeric(s, errors='coerce').astype('float64')

    rows = []
    for a, b in pairs:
        _sa_raw = df[a]
        _sb_raw = df[b]
        a_is_num = (a in _num_like)
        b_is_num = (b in _num_like)
        a_is_nom = (a in _nom_eligible)
        b_is_nom = (b in _nom_eligible)
        _sa = _encode_series(_sa_raw)
        _sb = _encode_series(_sb_raw)
        pair_df = pd.concat([_sa, _sb], axis=1).dropna()
        n = int(len(pair_df))
        if n < 2:
            r_val = np.nan; p_val = np.nan; dof = np.nan
        else:
            dof = n - 2
            # Correlation value (always Pearson on encoded data)
            if _have_scipy and a_is_num and b_is_num:
                try:
                    r_val, p_two = _sstats.pearsonr(pair_df.iloc[:,0].to_numpy(), pair_df.iloc[:,1].to_numpy())
                except Exception:
                    r_val = pair_df.iloc[:,0].corr(pair_df.iloc[:,1])
                    p_two = np.nan
            else:
                r_val = pair_df.iloc[:,0].corr(pair_df.iloc[:,1])
                p_two = np.nan

            # p-value policy:
            # - numeric↔numeric: use SciPy p (with alternative remap) when available, else NaN
            # - nominal↔nominal: KNIME shows p=0 → force 0.0
            if a_is_nom and b_is_nom:
                p_val = 0.0
            elif np.isnan(p_two):
                p_val = np.nan
            else:
                if _alt == 'TWO_SIDED':
                    p_val = float(p_two)
                elif _alt == 'GREATER':
                    p_val = float(p_two/2.0) if r_val >= 0 else float(1.0 - p_two/2.0)
                elif _alt == 'LESS':
                    p_val = float(p_two/2.0) if r_val <= 0 else float(1.0 - p_two/2.0)
                else:
                    p_val = float(p_two)

        rows.append({
            'First column name': a,
            'Second column name': b,
            'Correlation value': r_val,
            'p value': p_val,
            'Degrees of freedom': dof,
        })
    measure_out = pd.DataFrame(rows)
    # Do NOT sort — retain insertion order (matches KNIME)

# 5) Port 2 — numeric Pearson matrix in SOURCE order
_num_cols_ordered = [c for c in _union if c in _num_like]
if len(_num_cols_ordered) >= 1:
    _Xnum = df[_num_cols_ordered].apply(pd.to_numeric, errors='coerce').astype('float64')
    matrix_out = _Xnum.corr(method='pearson')
else:
    matrix_out = pd.DataFrame(index=pd.Index(_num_cols_ordered, name=None), columns=_num_cols_ordered, dtype='float64')

# 6) Port 3 — model
model_out = {
    'type': 'CorrelationModel',
    'method': 'pearson',
    'alternative': _alt if len(pairs) else 'TWO_SIDED',
    'columns': _union,
    'matrix': matrix_out.copy(),
}

port1 = measure_out
port2 = matrix_out
port3 = model_out
context['1552:1'] = port1
context['1552:2'] = port2
context['1552:3'] = port3
```
